```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python


df=pd.read_csv("../data/chicago_taxi_train.csv")
```

```python
df.head()
```

```python
df.describe()
```

```python
df["BASE FARE"]=df["FARE"]+df["TOLLS"]+df["EXTRAS"]
```

```python
df=df.drop(columns=["TRIP_START_HOUR","TRIP_SPEED","PICKUP_CENSUS_TRACT","DROPOFF_CENSUS_TRACT","PICKUP_COMMUNITY_AREA","DROPOFF_COMMUNITY_AREA","TIP_RATE","TIPS","TRIP_START_TIMESTAMP","TRIP_END_TIMESTAMP","PAYMENT_TYPE","COMPANY","TRIP_TOTAL"])
```

```python
X=df.drop(columns=["BASE FARE"])
Y=df["BASE FARE"]

```

```python
train_index=df.sample(frac=0.75,random_state=42).index
```

```python
X_train=X.loc[train_index]
Y_train=Y.loc[train_index]
```

```python
X_test=X.drop(train_index)
Y_test=Y.drop(train_index)
```

```python
X_train.to_csv("../data/X_train.csv")
Y_train.to_csv("../data/Y_train.csv")
X_test.to_csv("../data/X_test.csv")
Y_test.to_csv("../data/Y_test.csv")
```
