```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
X_train=pd.read_csv("../data/X_train.csv")
X_test=pd.read_csv("../data/X_test.csv")
Y_train=pd.read_csv("../data/Y_train.csv",usecols=[1])
Y_test=pd.read_csv("../data/Y_test.csv")
```

```python
w1=0
w2=0
b1=0
x1 = np.linspace(X_train["TRIP_SECONDS"].min(),
                 X_train["TRIP_SECONDS"].max(), 50)

x2 = np.linspace(X_train["TRIP_MILES"].min(),
                 X_train["TRIP_MILES"].max(), 50)

X1, X2 = np.meshgrid(x1, x2)
Z = w1 * X1 + w2 * X2 + b1
Z
```

```python
fig=plt.figure(figsize=(10,6))
ax=fig.add_subplot(111,projection="3d")
features=["TRIP_SECONDS","TRIP_MILES"]
ax.scatter(X_train[features[0]],X_train[features[1]],Y_train["BASE FARE"]) 
ax.set_xlabel("TRIP_SECONDS")
ax.set_ylabel("TRIP_MILES")
ax.set_zlabel("BASE_FARE")
ax.plot_surface(X1, X2, Z, alpha=0.5,color="pink")
ax.set_title("intial Linear Regression ")
```

```python
def predict(x1,x2):
    prediction=b1+w1*x1+w2*x2
    print("prediction of the LR model",prediction)


```
